```python
# Демонстрация работы с исходным иерархическим XML-файлом ЦБ РФ
import xml.etree.ElementTree as ET

# 1. Парсим сырой XML
tree = ET.parse('XML_daily.asp.xml')
root = tree.getroot()

print(f"Корневой тег: {root.tag}, Атрибуты: {root.attrib}")

# 2. Достаем данные из иерархии (например, USD и EUR)
xml_data = []
for valute in root.findall('Valute'):
    char_code = valute.find('CharCode').text
    if char_code in ['USD', 'EUR']:
        value = valute.find('Value').text.replace(',', '.')
        nominal = valute.find('Nominal').text
        xml_data.append({
            'Currency': char_code,
            'Nominal': nominal,
            'Value': float(value)
        })

print("\nДанные, извлеченные из XML:")
print(pd.DataFrame(xml_data))
```

# Источник с API с выдачей в формате JSON

```python
import requests
url = "https://archive-api.open-meteo.com/v1/archive"
params = {
    "latitude": 55.75, "longitude": 37.62,
    "start_date": "2020-01-01", "end_date": "2026-04-01",
    "daily": ["temperature_2m_mean", "precipitation_sum"],
    "timezone": "Europe/Moscow"
}
data = requests.get(url, params=params).json()
```
